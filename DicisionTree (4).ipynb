```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("/content/tennis_anyone_1000 (5).xlsx")
```

```python
df.head()
```

```python
df.columns
```

```python
df.drop(columns='Day ID',inplace=True)
```

```python
df.nunique()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Outlook']=le.fit_transform(df['Outlook']) #overcost-0,rain-1,sunny-2
df['Temperature']=le.fit_transform(df['Temperature'])#cool-0,mild-2,hot-1
df['Humidity']=le.fit_transform(df['Humidity'])#normal-0,high=1
df['Wind']=le.fit_transform(df['Wind'])#strong-0,weak-1
df['Play Tennis']=le.fit_transform(df['Play Tennis'])#yes=1,no-0
```

```python
df.head()
```

```python
### How to cheeck number of yes nad no rows in dataset

df.groupby('Play Tennis').count()
```

```python
### renaming the columns
df.rename(columns={'play Tennies':'playTannies'},inplace=True)
```

```python
df['Play Tennis'].value_counts()
```

```python
df.shape
```

```python
#### splitting dataset verticially(input/output)
x = df.iloc[:,0:-1]
y = df.iloc[:,-1]
```

```python
## splitting dataset horizontally(train and test set(20%))
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
### using Dicision Tree model
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()
dt
```

```python
# training the model
dt.fit(x_train,y_train)
```

```python
### predict

predict = dt.predict(x_test)
```

```python
## evaluating model
from sklearn.metrics import classification_report

print(classification_report(y_test,predict))
```

>>Dicision Tree accuracy with 100%
